```python
class Book:
  def __init__(self, title, author, is_borrowed = False):
    self.title = title
    self.author = author
    self.is_borrowed = is_borrowed
  def borrow(self):
    self.is_borrowed = True
    print(f"The book {self.title} by {self.author} has been borrowed!")
  def return_book(self):
    self.is_borrowed = False
    print(f"Kindly return the book {self.title} by {self.author}!")

Book1 = Book("Percy Jackson And The Lightning Thief", "Rick Riordan")
Book2 = Book("The Heroes of Olympus: The House of Hades", "Rick Riordan")
Book3 = Book("Omniscient Reader's Viewpoint", "SingNSong")

Book1.borrow()
Book1.return_book()
print("\n")
Book2.borrow()
Book2.return_book()
print("\n")
Book3.borrow()
Book3.return_book()
```
